堆叠

```python
a = np.arange(0, 4).reshape(2, 2)  # create a 2D array
b = np.arange(4, 8).reshape(2, 2)
print(a, end='\n\n')
print(b, end='\n\n')

print(np.vstack((a, b)), end='\n\n')  # stack arrays vertically
print(np.hstack((a, b)), end='\n\n')  # stack arrays horizontally
print(np.column_stack((a,b)), end='\n\n')

c=a.reshape(4)  # reshape to a column vector
d=b.reshape(4)  # reshape to a row vector
print(np.column_stack((c,d)), end='\n\n')
```

```python
a = np.arange(0, 4)
b = np.arange(4, 8)

print(a[:, np.newaxis], end='\n\n')  # add a new axis to a
print(np.column_stack((a[:], b[:])), end='\n\n')
print(np.hstack((a[:, np.newaxis], b[:, np.newaxis])), end='\n\n')  # the result is the same
```
